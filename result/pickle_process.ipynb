```python
import os
import pickle
import numpy as np

root_dir = './tune_and_exp'
output_dir = 'csv_outputs'
os.makedirs(output_dir, exist_ok=True)

for dirpath, dirnames, filenames in os.walk(root_dir):
    for filename in filenames:
        if filename == 'result.pkl':
            file_path = os.path.join(dirpath, filename)
            # Create a unique name based on the folder structure
            rel_path = os.path.relpath(dirpath, root_dir).replace(os.sep, '_')
            val_csv = os.path.join(output_dir, f'{rel_path}_acc_array_val.csv')
            test_csv = os.path.join(output_dir, f'{rel_path}_acc_array_test.csv')
            try:
                with open(file_path, 'rb') as f:
                    data = pickle.load(f)
                val_accuracies = data.get('acc_array_val')
                test_accuracies = data.get('acc_array_test')
                if val_accuracies is not None:
                    np.savetxt(val_csv, val_accuracies.reshape(-1, val_accuracies.shape[-1]), delimiter=',')
                if test_accuracies is not None:
                    np.savetxt(test_csv, test_accuracies.reshape(-1, test_accuracies.shape[-1]), delimiter=',')
                print(f"Saved: {val_csv}, {test_csv}")
            except Exception as e:
                print(f"Failed to process {file_path}: {e}")
```
