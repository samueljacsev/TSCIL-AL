```python
uwave_ude = np.array([92.14, 94.03, 96.29, 93.53])
uwave_ri = np.array([90.49, 92.7, 93.19, 90.27])

har_ude = np.array([94.7, 98.41, 96.44, 98.16])
har_ri = np.array([94.5, 95.27, 94.28, 96.62])
```

```python
diffs = har_ude - har_ri

# Paired t-test
stat, p = ttest_rel(har_ude, har_ri)

# Paired t-test
mean_diff = np.mean(diffs)
std_diff = np.std(diffs, ddof=1)  # sample std
sem_diff = std_diff / np.sqrt(len(diffs))  # standard error of the mean

{
    'differences': diffs.tolist(),
    'mean_difference': mean_diff,
    'std_difference': std_diff,
    'sem_difference': sem_diff,
    't_statistic': stat,
    'p_value': p
}
```
