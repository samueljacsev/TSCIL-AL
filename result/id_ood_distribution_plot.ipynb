# DATASET: HAR 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_id = pd.read_csv("run0_task1_msp_id.csv")
df_cur = pd.read_csv("run0_task1_msp_current.csv")

plt.figure(figsize=(7,4))
sns.kdeplot(df_id["score"], label="Previous ID", fill=True, alpha=0.3)
sns.kdeplot(df_cur[df_cur["true_label"]==0]["score"], label="Current ID", fill=True, alpha=0.3)
sns.kdeplot(df_cur[df_cur["true_label"]==1]["score"], label="Current OOD", fill=True, alpha=0.3)
plt.axvline(df_cur["threshold"].iloc[0], color='red', linestyle='--', label='Threshold')
plt.legend()
plt.title("OOD Score Distributions - Task 1, MSP")
plt.show()
```
