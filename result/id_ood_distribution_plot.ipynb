## OOD performance 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import os
from pathlib import Path

# =============================================================================
# PART 1: DATA EXTRACTION FROM LOG FILES
# =============================================================================

def extract_ood_metrics_from_log(log_file_path):
    """
    Extract OOD detection metrics from a log file.
    Returns a list of dictionaries with metrics for each task.
    """
    with open(log_file_path, 'r', encoding='utf-8', errors='ignore') as f:
        content = f.read()
    
    # Split by runs
    run_sections = re.split(r'######## Run (\d+)', content)
    
    metrics_list = []
    
    for i in range(1, len(run_sections), 2):
        run_num = int(run_sections[i])
        run_content = run_sections[i + 1]
        
        # Find all task sections
        task_pattern = r'\[Eval\] Task (\d+) \| ROC AUC: ([\d.]+) \| est_ratio: ([\d.]+)'
        confusion_pattern = r'\[Confusion\] TP=(\d+), FN=(\d+), FP=(\d+), TN=(\d+)'
        
        tasks = re.finditer(task_pattern, run_content)
        confusions = re.finditer(confusion_pattern, run_content)
        
        task_list = list(tasks)
        confusion_list = list(confusions)
        
        for task_match, confusion_match in zip(task_list, confusion_list):
            task_id = int(task_match.group(1))
            roc_auc = float(task_match.group(2))
            est_ratio = float(task_match.group(3))
            
            tp = int(confusion_match.group(1))
            fn = int(confusion_match.group(2))
            fp = int(confusion_match.group(3))
            tn = int(confusion_match.group(4))
            
            # Calculate metrics
            precision = tp / (tp + fp) if (tp + fp) > 0 else 0
            recall = tp / (tp + fn) if (tp + fn) > 0 else 0
            f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
            accuracy = (tp + tn) / (tp + tn + fp + fn) if (tp + tn + fp + fn) > 0 else 0
            
            metrics_list.append({
                'run': run_num,
                'task': task_id,
                'roc_auc': roc_auc,
                'est_ratio': est_ratio,
                'tp': tp,
                'fn': fn,
                'fp': fp,
                'tn': tn,
                'precision': precision,
                'recall': recall,
                'f1_score': f1_score,
                'accuracy': accuracy
            })
    
    return metrics_list

# Define log file mapping
log_files = {
    ('dailysports', 'knn'): r'tune_and_exp\CNN_dailysports\ASER_BN_Nov-19-19-42-26\log.txt',
    ('dailysports', 'cosine'): r'tune_and_exp\CNN_dailysports\ASER_BN_Nov-19-19-44-02\log.txt',
    ('har', 'knn'): r'tune_and_exp\CNN_har\ASER_BN_Nov-19-19-45-49\log.txt',
    ('har', 'cosine'): r'tune_and_exp\CNN_har\ASER_BN_Nov-19-19-46-38\log.txt',
    ('uwave', 'knn'): r'tune_and_exp\CNN_uwave\ASER_BN_Nov-18-21-31-11\log.txt',
    ('uwave', 'cosine'): r'tune_and_exp\CNN_uwave\ASER_BN_Nov-18-21-32-36\log.txt',
}

# Extract all metrics
all_metrics = []

for (dataset, method), log_path in log_files.items():
    print(f"Processing {dataset} - {method}...")
    try:
        metrics = extract_ood_metrics_from_log(log_path)
        for m in metrics:
            m['dataset'] = dataset
            m['method'] = method
        all_metrics.extend(metrics)
        print(f"  Extracted {len(metrics)} task evaluations")
    except Exception as e:
        print(f"  ERROR: {e}")

# Create DataFrame
df_ood = pd.DataFrame(all_metrics)
print(f"\nTotal extracted records: {len(df_ood)}")
print("\nDataFrame preview:")
print(df_ood.head(10))

# =============================================================================
# PART 2: AGGREGATE METRICS (Averaged across all tasks and 3 runs)
# =============================================================================

print("\n" + "="*80)
print("AGGREGATE OOD PERFORMANCE (Averaged across all tasks and 3 runs)")
print("="*80)

aggregate_metrics = df_ood.groupby(['dataset', 'method']).agg({
    'roc_auc': ['mean', 'std'],
    'precision': ['mean', 'std'],
    'recall': ['mean', 'std'],
    'f1_score': ['mean', 'std'],
    'accuracy': ['mean', 'std']
}).round(4)

print(aggregate_metrics)

# =============================================================================
# PART 3: AGGREGATE CONFUSION MATRIX (Summed across all tasks and 3 runs)
# =============================================================================

print("\n" + "="*80)
print("AGGREGATE CONFUSION MATRIX (Summed across all tasks and 3 runs)")
print("="*80)

confusion_aggregates = df_ood.groupby(['dataset', 'method']).agg({
    'tp': 'sum',
    'fn': 'sum',
    'fp': 'sum',
    'tn': 'sum'
}).astype(int)

print(confusion_aggregates)
```

```python
# =============================================================================
# PART 4: VISUALIZATIONS
# =============================================================================

# Set publication-quality style
plt.style.use('seaborn-v0_8-paper')
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'DejaVu Serif'],
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 13,
    'figure.dpi': 300,
})

# Color palette
method_colors = {'cosine': '#0173B2', 'knn': '#DE8F05'}
dataset_order = ['uwave', 'har', 'dailysports']
dataset_names = {'uwave': 'UWave', 'har': 'UCI HAR', 'dailysports': 'Daily Sports'}

# -----------------------------------------------------------------------------
# PLOT 1: Per-Dataset OOD Performance - ROC AUC ONLY (Bar Chart)
# -----------------------------------------------------------------------------

print("\nGenerating Plot 1: Per-Dataset ROC AUC Performance...")

fig, ax = plt.subplots(1, 1, figsize=(10, 6))

# Prepare data for plotting
plot_data = df_ood.groupby(['dataset', 'method'])['roc_auc'].mean().unstack()
plot_data = plot_data.reindex(dataset_order)

x = np.arange(len(dataset_order))
width = 0.35

for i, method in enumerate(['cosine', 'knn']):
    values = [plot_data.loc[ds, method] if method in plot_data.columns else 0 
              for ds in dataset_order]
    ax.bar(x + i*width, values, width, label=method.capitalize(), 
           color=method_colors[method], alpha=0.85)

ax.set_xlabel('Dataset', fontsize=12, fontweight='bold')
ax.set_ylabel('ROC AUC', fontsize=12, fontweight='bold')
ax.set_title('OOD Detection Performance (ROC AUC)', fontsize=14, fontweight='bold', pad=15)
ax.set_xticks(x + width / 2)
ax.set_xticklabels([dataset_names[ds] for ds in dataset_order])
ax.legend(loc='best', frameon=True, edgecolor='black', fontsize=11)
ax.grid(True, alpha=0.3, axis='y', linestyle='--')
ax.set_ylim([0, 1.0])

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('plots/ood_roc_auc_bar_chart.png', dpi=300, bbox_inches='tight')
plt.savefig('plots/ood_roc_auc_bar_chart.pdf', dpi=300, bbox_inches='tight')
print("  Saved: plots/ood_roc_auc_bar_chart.png and .pdf")
plt.show()

# -----------------------------------------------------------------------------
# PLOT 2: Confusion Matrices (Heatmaps)
# -----------------------------------------------------------------------------

print("\nGenerating Plot 2: Confusion Matrices...")

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for idx, dataset in enumerate(dataset_order):
    for jdx, method in enumerate(['cosine', 'knn']):
        ax = axes[jdx, idx]
        
        # Get confusion matrix data
        subset = df_ood[(df_ood['dataset'] == dataset) & (df_ood['method'] == method)]
        
        if len(subset) == 0:
            ax.text(0.5, 0.5, 'No Data', ha='center', va='center')
            ax.set_title(f'{dataset_names[dataset]} - {method.capitalize()}')
            continue
        
        # Sum confusion matrix across all tasks and runs
        tp = subset['tp'].sum()
        fn = subset['fn'].sum()
        fp = subset['fp'].sum()
        tn = subset['tn'].sum()
        
        # Create confusion matrix
        cm = np.array([[tn, fp], [fn, tp]])
        
        # Normalize by row (true labels)
        cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        
        # Plot heatmap
        sns.heatmap(cm_norm, annot=True, fmt='.2%', cmap='Blues', 
                   cbar=True, ax=ax, vmin=0, vmax=1,
                   xticklabels=['Predicted ID', 'Predicted OOD'],
                   yticklabels=['True ID', 'True OOD'])
        
        # Add raw counts as text
        for i in range(2):
            for j in range(2):
                ax.text(j+0.5, i+0.7, f'({cm[i,j]})', 
                       ha='center', va='center', fontsize=9, color='gray')
        
        ax.set_title(f'{dataset_names[dataset]} - {method.capitalize()}', 
                    fontsize=12, fontweight='bold', pad=10)
        ax.set_xlabel('Predicted Label', fontsize=11)
        ax.set_ylabel('True Label', fontsize=11)

plt.tight_layout()
plt.savefig('plots/ood_confusion_matrices.png', dpi=300, bbox_inches='tight')
plt.savefig('plots/ood_confusion_matrices.pdf', dpi=300, bbox_inches='tight')
print("  Saved: plots/ood_confusion_matrices.png and .pdf")
plt.show()
```

```python
# -----------------------------------------------------------------------------
# PLOT 3: Per-Task Performance - ROC AUC ONLY (Line Chart)
# -----------------------------------------------------------------------------

print("\nGenerating Plot 3: Per-Task ROC AUC Performance...")

# Average across runs for each dataset, method, task combination
df_task_avg = df_ood.groupby(['dataset', 'method', 'task']).agg({
    'roc_auc': 'mean',
}).reset_index()

# Single plot with all datasets and methods
fig, ax = plt.subplots(1, 1, figsize=(12, 6))

for dataset in dataset_order:
    for method in ['cosine', 'knn']:
        subset = df_task_avg[(df_task_avg['dataset'] == dataset) & 
                             (df_task_avg['method'] == method)]
        
        if len(subset) == 0:
            continue
        
        # Sort by task
        subset = subset.sort_values('task')
        
        linestyle = '-' if method == 'cosine' else '--'
        marker = 'o' if method == 'cosine' else 's'
        
        label_text = f'{dataset_names[dataset]} - {method.capitalize()}'
        ax.plot(subset['task'], subset['roc_auc'], 
               marker=marker, linestyle=linestyle, linewidth=2, markersize=6,
               label=label_text, alpha=0.8)

ax.set_xlabel('Task Number', fontsize=12, fontweight='bold')
ax.set_ylabel('ROC AUC', fontsize=12, fontweight='bold')
ax.set_title('OOD Detection ROC AUC per Task', fontsize=14, fontweight='bold', pad=15)
ax.legend(loc='best', frameon=True, edgecolor='black', fontsize=10, ncol=2)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_ylim([0, 1.0])

# Set x-axis to show only integer values
ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('plots/ood_roc_auc_per_task.png', dpi=300, bbox_inches='tight')
plt.savefig('plots/ood_roc_auc_per_task.pdf', dpi=300, bbox_inches='tight')
print("  Saved: plots/ood_roc_auc_per_task.png and .pdf")
plt.show()

# -----------------------------------------------------------------------------
# PLOT 4: Alternative Per-Task View - Separate Subplots per Dataset (ROC AUC)
# -----------------------------------------------------------------------------

print("\nGenerating Plot 4: Per-Task Performance by Dataset (separate panels)...")

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, dataset in enumerate(dataset_order):
    ax = axes[idx]
    
    for method in ['cosine', 'knn']:
        subset = df_task_avg[(df_task_avg['dataset'] == dataset) & 
                             (df_task_avg['method'] == method)]
        
        if len(subset) == 0:
            continue
        
        # Sort by task
        subset = subset.sort_values('task')
        
        linestyle = '-' if method == 'cosine' else '--'
        marker = 'o' if method == 'cosine' else 's'
        
        # Plot ROC AUC (primary metric)
        ax.plot(subset['task'], subset['roc_auc'], 
               marker=marker, linestyle=linestyle, linewidth=2.5, markersize=8,
               label=f'{method.capitalize()}', color=method_colors[method], alpha=0.85)
    
    ax.set_xlabel('Task Number', fontsize=12, fontweight='bold')
    ax.set_ylabel('ROC AUC', fontsize=12, fontweight='bold')
    ax.set_title(f'{dataset_names[dataset]}', fontsize=14, fontweight='bold', pad=10)
    ax.legend(loc='best', frameon=True, edgecolor='black', fontsize=11)
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.set_ylim([0, 1.0])
    
    # Set x-axis to show only integer values
    ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
    
    # Remove top and right spines
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('plots/ood_roc_auc_per_task_by_dataset.png', dpi=300, bbox_inches='tight')
plt.savefig('plots/ood_roc_auc_per_task_by_dataset.pdf', dpi=300, bbox_inches='tight')
print("  Saved: plots/ood_roc_auc_per_task_by_dataset.png and .pdf")
plt.show()

# =============================================================================
# PART 5: DETAILED ANALYSIS FOR THESIS
# =============================================================================

print("\n" + "="*80)
print("DETAILED ANALYSIS FOR THESIS - LaTeX FORMAT")
print("="*80)

# ============================================================================
# Analysis 1: Overall Performance Summary with Statistical Significance
# ============================================================================

print("\n" + "-"*80)
print("1. OVERALL OOD DETECTION PERFORMANCE SUMMARY")
print("-"*80)

overall_stats = df_ood.groupby(['dataset', 'method']).agg({
    'roc_auc': ['mean', 'std', 'min', 'max', 'count'],
    'precision': ['mean', 'std'],
    'recall': ['mean', 'std'],
    'f1_score': ['mean', 'std']
}).round(4)

print("\nRaw Statistics:")
print(overall_stats)

# Per-method comparison
print("\n\nMETHOD COMPARISON (Averaged across all datasets):")
print("-" * 60)
method_comparison = df_ood.groupby('method').agg({
    'roc_auc': ['mean', 'std'],
    'precision': ['mean', 'std'],
    'recall': ['mean', 'std'],
    'f1_score': ['mean', 'std']
}).round(4)
print(method_comparison)

# Per-dataset comparison
print("\n\nDATASET COMPARISON (Averaged across both methods):")
print("-" * 60)
dataset_comparison = df_ood.groupby('dataset').agg({
    'roc_auc': ['mean', 'std'],
    'precision': ['mean', 'std'],
    'recall': ['mean', 'std'],
    'f1_score': ['mean', 'std']
}).round(4)
print(dataset_comparison)

# ============================================================================
# Analysis 2: Per-Dataset Detailed Results
# ============================================================================

print("\n" + "-"*80)
print("2. PER-DATASET DETAILED ANALYSIS")
print("-"*80)

for dataset in dataset_order:
    print(f"\n{'='*60}")
    print(f"DATASET: {dataset_names[dataset].upper()}")
    print(f"{'='*60}")
    
    dataset_df = df_ood[df_ood['dataset'] == dataset]
    
    for method in ['cosine', 'knn']:
        method_df = dataset_df[dataset_df['method'] == method]
        
        print(f"\n  Method: {method.upper()}")
        print(f"  {'-'*50}")
        print(f"    ROC AUC:   {method_df['roc_auc'].mean():.4f} ± {method_df['roc_auc'].std():.4f}")
        print(f"    Precision: {method_df['precision'].mean():.4f} ± {method_df['precision'].std():.4f}")
        print(f"    Recall:    {method_df['recall'].mean():.4f} ± {method_df['recall'].std():.4f}")
        print(f"    F1 Score:  {method_df['f1_score'].mean():.4f} ± {method_df['f1_score'].std():.4f}")
        print(f"    # Evaluations: {len(method_df)}")
    
    # Compare methods for this dataset
    cosine_auc = dataset_df[dataset_df['method'] == 'cosine']['roc_auc'].mean()
    knn_auc = dataset_df[dataset_df['method'] == 'knn']['roc_auc'].mean()
    improvement = ((cosine_auc - knn_auc) / knn_auc) * 100 if knn_auc > 0 else 0
    
    better_method = 'Cosine' if cosine_auc > knn_auc else 'kNN'
    print(f"\n  Winner: {better_method}")
    print(f"  Cosine vs kNN: {cosine_auc:.4f} vs {knn_auc:.4f}")
    print(f"  Relative improvement: {abs(improvement):.2f}% ({'Cosine' if improvement > 0 else 'kNN'} better)")

# ============================================================================
# Analysis 3: Confusion Matrix Analysis
# ============================================================================

print("\n" + "-"*80)
print("3. CONFUSION MATRIX ANALYSIS")
print("-"*80)

print("\nAggregate Confusion Matrices (All tasks & runs combined):")
print(confusion_aggregates)

print("\n\nDerived Metrics from Confusion Matrices:")
print("-" * 60)

for (dataset, method), row in confusion_aggregates.iterrows():
    tp, fn, fp, tn = row['tp'], row['fn'], row['fp'], row['tn']
    
    # Calculate metrics
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    accuracy = (tp + tn) / (tp + tn + fp + fn) if (tp + tn + fp + fn) > 0 else 0
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0  # False Positive Rate
    fnr = fn / (fn + tp) if (fn + tp) > 0 else 0  # False Negative Rate
    
    print(f"\n{dataset_names[dataset]} - {method.upper()}:")
    print(f"  True Positives:  {tp:5d}  |  False Positives: {fp:5d}")
    print(f"  False Negatives: {fn:5d}  |  True Negatives:  {tn:5d}")
    print(f"  Precision:   {precision:.4f}  (How many detected OOD are actually OOD)")
    print(f"  Recall:      {recall:.4f}  (How many actual OOD were detected)")
    print(f"  F1 Score:    {f1:.4f}")
    print(f"  Accuracy:    {accuracy:.4f}")
    print(f"  Specificity: {specificity:.4f}  (True Negative Rate)")
    print(f"  FPR:         {fpr:.4f}  (False alarms on ID samples)")
    print(f"  FNR:         {fnr:.4f}  (Missed OOD samples)")

# ============================================================================
# Analysis 4: Task Progression Analysis
# ============================================================================

print("\n" + "-"*80)
print("4. TASK PROGRESSION ANALYSIS")
print("-"*80)

print("\nHow does OOD detection performance change across tasks?")
print("-" * 60)

for dataset in dataset_order:
    print(f"\n{dataset_names[dataset].upper()}:")
    dataset_task_df = df_task_avg[df_task_avg['dataset'] == dataset]
    
    for method in ['cosine', 'knn']:
        method_task_df = dataset_task_df[dataset_task_df['method'] == method].sort_values('task')
        
        if len(method_task_df) == 0:
            continue
        
        tasks = method_task_df['task'].values
        aucs = method_task_df['roc_auc'].values
        
        print(f"\n  {method.upper()}:")
        for task, auc in zip(tasks, aucs):
            print(f"    Task {task}: {auc:.4f}")
        
        # Trend analysis
        if len(aucs) > 1:
            first_task_auc = aucs[0]
            last_task_auc = aucs[-1]
            trend = "improving" if last_task_auc > first_task_auc else "degrading"
            change = ((last_task_auc - first_task_auc) / first_task_auc) * 100
            
            print(f"    Trend: {trend} ({change:+.2f}% from Task {tasks[0]} to Task {tasks[-1]})")
            print(f"    Mean: {aucs.mean():.4f}, Std: {aucs.std():.4f}")
            print(f"    Min: {aucs.min():.4f} (Task {tasks[aucs.argmin()]}), Max: {aucs.max():.4f} (Task {tasks[aucs.argmax()]})")

# ============================================================================
# Analysis 5: LaTeX Table Generation
# ============================================================================

print("\n" + "="*80)
print("5. LaTeX TABLE GENERATION")
print("="*80)

# Table 1: Overall Results
print("\n\\begin{table}[htbp]")
print("\\centering")
print("\\caption{Overall OOD Detection Performance (Mean ± Std)}")
print("\\label{tab:ood_overall}")
print("\\begin{tabular}{llcccc}")
print("\\hline")
print("\\textbf{Dataset} & \\textbf{Method} & \\textbf{ROC AUC} & \\textbf{Precision} & \\textbf{Recall} & \\textbf{F1 Score} \\\\")
print("\\hline")

for dataset in dataset_order:
    dataset_df = df_ood[df_ood['dataset'] == dataset]
    
    for method in ['cosine', 'knn']:
        method_df = dataset_df[dataset_df['method'] == method]
        
        roc_mean = method_df['roc_auc'].mean()
        roc_std = method_df['roc_auc'].std()
        prec_mean = method_df['precision'].mean()
        prec_std = method_df['precision'].std()
        rec_mean = method_df['recall'].mean()
        rec_std = method_df['recall'].std()
        f1_mean = method_df['f1_score'].mean()
        f1_std = method_df['f1_score'].std()
        
        method_display = "Cosine" if method == 'cosine' else "kNN"
        
        print(f"{dataset_names[dataset]} & {method_display} & "
              f"{roc_mean:.3f}$\\pm${roc_std:.3f} & "
              f"{prec_mean:.3f}$\\pm${prec_std:.3f} & "
              f"{rec_mean:.3f}$\\pm${rec_std:.3f} & "
              f"{f1_mean:.3f}$\\pm${f1_std:.3f} \\\\")
    
    if dataset != dataset_order[-1]:
        print("\\hline")

print("\\hline")
print("\\end{tabular}")
print("\\end{table}")

# Table 2: Confusion Matrix Summary
print("\n\n\\begin{table}[htbp]")
print("\\centering")
print("\\caption{Aggregate Confusion Matrix Results}")
print("\\label{tab:ood_confusion}")
print("\\begin{tabular}{llcccc}")
print("\\hline")
print("\\textbf{Dataset} & \\textbf{Method} & \\textbf{TP} & \\textbf{FN} & \\textbf{FP} & \\textbf{TN} \\\\")
print("\\hline")

for (dataset, method), row in confusion_aggregates.iterrows():
    method_display = "Cosine" if method == 'cosine' else "kNN"
    print(f"{dataset_names[dataset]} & {method_display} & "
          f"{row['tp']} & {row['fn']} & {row['fp']} & {row['tn']} \\\\")
    
print("\\hline")
print("\\end{tabular}")
print("\\end{table}")

# ============================================================================
# Analysis 6: Key Findings Summary
# ============================================================================

print("\n" + "="*80)
print("6. KEY FINDINGS SUMMARY FOR THESIS")
print("="*80)

# Find best performing method overall
overall_method_perf = df_ood.groupby('method')['roc_auc'].mean()
best_method = overall_method_perf.idxmax()
best_method_auc = overall_method_perf.max()
worst_method_auc = overall_method_perf.min()

print(f"\n✓ Best Overall Method: {best_method.upper()}")
print(f"  - Mean ROC AUC: {best_method_auc:.4f}")
print(f"  - Improvement over {('kNN' if best_method == 'cosine' else 'Cosine')}: "
      f"{((best_method_auc - worst_method_auc) / worst_method_auc * 100):.2f}%")

# Find best dataset for OOD detection
overall_dataset_perf = df_ood.groupby('dataset')['roc_auc'].mean()
best_dataset = overall_dataset_perf.idxmax()
worst_dataset = overall_dataset_perf.idxmin()

print(f"\n✓ Easiest Dataset for OOD Detection: {dataset_names[best_dataset].upper()}")
print(f"  - Mean ROC AUC: {overall_dataset_perf[best_dataset]:.4f}")

print(f"\n✓ Most Challenging Dataset: {dataset_names[worst_dataset].upper()}")
print(f"  - Mean ROC AUC: {overall_dataset_perf[worst_dataset]:.4f}")

# Consistency analysis
print("\n✓ Consistency Analysis (based on std deviation):")
for method in ['cosine', 'knn']:
    method_std = df_ood[df_ood['method'] == method]['roc_auc'].std()
    print(f"  - {method.upper()}: σ = {method_std:.4f}")

print("\n" + "="*80)
print("ANALYSIS COMPLETE - ALL RESULTS SAVED")
print("="*80)

# Export detailed results to CSV
detailed_results = df_ood.groupby(['dataset', 'method']).agg({
    'roc_auc': ['mean', 'std', 'min', 'max'],
    'precision': ['mean', 'std'],
    'recall': ['mean', 'std'],
    'f1_score': ['mean', 'std']
}).round(4)

detailed_results.to_csv('plots/ood_detailed_analysis.csv')
print("\n✓ Detailed analysis exported to: plots/ood_detailed_analysis.csv")

# SUMMARY TABLES FOR THESIS

print("\n" + "="*80)
print("SUMMARY TABLES FOR THESIS")
print("="*80)

# Table 1: Overall Performance Summary (ROC AUC focused)
print("\nTable 1: Overall OOD Detection Performance (Mean ± Std across all tasks and runs)")
print("-"*80)

summary_table = df_ood.groupby(['dataset', 'method']).agg({
    'roc_auc': lambda x: f"{x.mean():.3f} ± {x.std():.3f}",
    'precision': lambda x: f"{x.mean():.3f} ± {x.std():.3f}",
    'recall': lambda x: f"{x.mean():.3f} ± {x.std():.3f}",
    'f1_score': lambda x: f"{x.mean():.3f} ± {x.std():.3f}",
})
summary_table.columns = ['ROC AUC', 'Precision', 'Recall', 'F1 Score']
print(summary_table)

# Table 2: Confusion Matrix Summary
print("\n\nTable 2: Aggregate Confusion Matrix (Total counts across all tasks and runs)")
print("-"*80)
print(confusion_aggregates)

# Export to CSV
summary_table.to_csv('plots/ood_performance_summary.csv')
confusion_aggregates.to_csv('plots/ood_confusion_summary.csv')
print("\n✓ Tables exported to CSV files in plots/ folder")

print("\n" + "="*80)
print("ALL OOD ANALYSIS COMPLETE!")
print("="*80)
print("\nGenerated files:")
print("  1. plots/ood_roc_auc_bar_chart.png/pdf (ROC AUC bar chart)")
print("  2. plots/ood_confusion_matrices.png/pdf (Confusion matrices)")
print("  3. plots/ood_roc_auc_per_task.png/pdf (ROC AUC line chart - all datasets)")
print("  4. plots/ood_roc_auc_per_task_by_dataset.png/pdf (ROC AUC by dataset)")
print("  5. plots/ood_performance_summary.csv")
print("  6. plots/ood_confusion_summary.csv")
print("  7. plots/ood_detailed_analysis.csv (NEW - Detailed statistics)")
```
